Slicing & Reversal Write a function reverse_and_alternate(s) that takes a string s, reverses it completely, and then returns every second character of that reversed string.Example: "datascience"  reversed is "ecneicsatad"  returns "enisa"


```python
def stringfun(s):
    print(s,"This is orignal String")

    print(f"This is reversed string {s[::-1]}")

    print(f"Every second later {s[0::2]}")

name="Mayuresh"
stringfun(name)
```

### Problem 2:Character Frequency Count

Write a function "char_count(text)" that takes a string and returns a dictionary containing the frequency count of each character. Ignore spaces and make it case-insensitive.

Example: "Data Science"  
 {'d': 1, 'a': 2, 't': 1, 's': 1, 'c': 2, 'i': 1, 'e': 2, 'n': 1}

```python
def char_count(text):
    c={}
    for i in text:
        if i not in c:
            c[i]=1
        else:
            c[i]+=1
    print(c)




text=input("Enter your string :")
char_count(text)

```

## Problem 3: Second Largest Element
Write a function second_largest(numbers) that finds the second largest number in a list without using Python's built-in sort() or sorted() functions.

Example: [12, 45, 2, 45, 8, 33] → returns 33

```python
def second_largest(numbers):
    largest=0
    s_largest=0

    for n in numbers:
        if n > largest:
            s_largest=largest
            largest=n

        elif n != largest and  s_largest is None or n > s_largest:
            s_largest=n

    return s_largest

num=[20,89,67,56,56,78,89,90,93]
second_largest(num)
```

# Problem 4: List Flattening & Deduplication
Given a list of lists containing integers, write a function flatten_unique(nested_list) that flattens it into a single list of unique values, sorted in ascending order.

Example: [[1, 2, 3], [2, 4, 5], [1, 5, 6]] $\rightarrow$ returns [1, 2, 3, 4, 5, 6]

```python
def flatten_unique(nested_list):

  flatten_list=[]

  for number in nested_list:

    for n in number:

      if n not in flatten_list:
        flatten_list.append(n)

  return sorted(flatten_list)

n_list=[[23,34,45,5,6],[7,5,6,7,8],[6,5,4,3,20]]

result=flatten_unique(n_list)

print("The flatten list is:",result)

```
